## What is a Python package

A Distribution Package (aka Package) is a versioned archive file that contains Python modules and other resource files that are used to distribute a Release.
<br>
The archive file is what an end-user will download and install.

## Install a Python Package
How to install Python packages:
* Python packages can be installed directly from **source code**
* **OS package manager** (apt, yum, brew, ...) can install Python package
* **`pip`** is generally used to install Python packages
* **`conda`** is package management tool that can install packages provided by Anaconda on the `repo.continuum.io` repository.<br>
  It is becoming very popular especially for data science.

## pip

pip is the de facto package manager of Python language.<br>
The most common usage of pip is to install package taken from "Python Package Index" (PyPI) repository
(see https://pypi.org).<br>
PyPI is a public repository of open source licensed Python packages useful to install and shared software developed by Python community.

## pip

Before install packages pip:
* resolves the dependencies
* checks if they are already installed on the system, if not install them
* once all dependencies are satisfied, installs the requested package

## Where install packages

Python packages are installed in a standard location for third-party Python modules which is a specific directory called `site-packages` (`dist-package`).<br>
The location of `site-packages` directory varies by platform, by how Python is installed and by how the package is installed.

There are two types of `site-packages` directories: **global** and **user**

### Global site-packages

Global `site-packages` directories are listed in **`sys.path`**<br>
`sys.path` is a list of strings that specifies the search path for modules.<br>
`sys.path` is populated using the current working directory, followed by directories listed in `PYTHONPATH` environment variable, followed by installation-dependent default paths.

````python
import sys
sys.path
````

### User site-packages

User site-packages directories allow users to a local installation of Python packages in their home directory.<br>
see `PEP 370` (NB: Python Enhancement Proposals aka **PEP**)
<br>
https://www.python.org/dev/peps/pep-0370/ <br>

* Unix: `~/.local/lib/python3.6/site-packages`
* Windows: `%APPDATA%/Python/Python36/site-packages`

## Issues with global installations

* non-root user do not have permission to install in the global `site-packages` directory
* only a single version of a package can be installed for a given Python interpreter.<br>
  This causes issues when a package has dependencies from multiple libraries or applications
* upgrading the dependency, eg installing another package, will break applications or libraries
* Linux package managers (apt, yum, brew, ...) install in a non user writable location



## pip user mode

A "user scheme" for installation is introduced from Python 2.6.<br>
All Python distributions support an alternative install location specific to the user.<br>
To activate "user scheme" and install a package isolated to the current user use `--user` flag with pip:

pip install --user BeautifulProject

An user installation prevents breaking any system-wide packages.

## Virtual Environment

An alternative to user mode in pip is to use a Python **virtual environment** to install a Python project.<br>
    A virtual environment is a self confinement Python packages installation that can coexist with other installations on the same system.<br>
Thus an application can not share dependency with other application.



## `venv` Module

`venv` is a module to create isolated Python environments. \
`venv` creates a directory which contains all executables, library files and package manager to use Python packages.

From Python 3.3 the `venv` package has added to the standard library.

## virtual environment basic usage

1. Create a virtual environment for a project
1. Activate virtual environment
1. Install packages and/or Execute commands
1. When done deactivate virtual environment


## Create a virtual environment

````python
%%bash
python -m venv my_venv
````

This command creates a directory "`my_venv`" in which all python files are installed.

````python
%%bash
ls my_venv
````

By default `venv` module does not include global Python `site-packages` directory. \
To create a virtual environment including global `site-packages` directory:

````python
%%bash
python -m venv --system-site-packages my_venv
````

## Activate virtual environment

A virtual environment needs to be activated before using it:

````python
%%bash
source my_venv/bin/activate  # activate virtualenv
````

activate script create a subshell and add `bin` directory of virtual env to the `PATH` environment variable.<br>
Thus python, pip, ... executables are taken from this directory and not from one installed globally on the system.

The name of the environment will appear as shell prompt prefix on the left of the original prompt, e.g.:<br>
`(my_venv) $`

On Windows:

my_venv\Scripts\activate.bat

## Install package in a virtual environment

%%bash
pip install BeautifulPackage

Install package in `site-packages` directory of virtual environment.

## Deactivate a virtual environment

When done you have to deactivate virtual environment:

%%bash
deactivate

This returns you back to the system’s default Python interpreter with all its installed libraries.

## Environment common usage

````python
%%bash

python -m venv my_venv        # create virtual env

source my_venv/bin/activate   # activate virtualenv

# 
# some useful commands
#
pip list

deactivate                    # exit from a virtualenv
````

## The requirement file

If you want to specify all dependencies and versions used in a project you can use a **requirement file**.<br>
A requirement file allows to specify which packages and versions should be installed for the project.<br>
`pip freeze` command shows all installed packages in requirements format.<br>
Its output can be redirect to a file (the requirement file).

````python
pip freeze > requirements.txt
````

Once you have a requirement file you can replicate the same environment in another system with the command `pip install` with `-r` option:

````python
pip install -r requirements.txt
````

## uv

`uv` is a fast Python package and environment manager developed by Astral.<br>

It is designed as a modern replacement for tools like `pip`, `pip-tools`, and sometimes even virtualenv managers, with a strong focus on **speed** and **simple workflows**.<br>
`uv` can create and manage virtual environments, install and resolve dependencies, and run Python commands using isolated environments.<br>
It integrates well with existing Python projects and can read standard files like `pyproject.toml` and `requirements.txt`.

Some typical use cases:

* **Project environments**: quickly create and manage isolated environments for your Python projects.
* **Dependency management**: install, upgrade, and lock dependencies very efficiently.
* **Running tools**: use `uv run` to execute Python scripts and CLI tools in a temporary or project-specific environment without manual setup.


### Example: using `uv` for a small project

```bash
uv init example

cd example

# Add a dependency (ruff)
uv add ruff

# Run ruff inside the uv-managed environment
uv run ruff check

# Generate a lockfile
uv lock

# Sync the environment from the lockfile
uv sync
