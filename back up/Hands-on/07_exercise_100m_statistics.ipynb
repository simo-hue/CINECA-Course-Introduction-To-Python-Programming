# Men's 100 meters world record statistics

```python
file_data = "times_100m.dat"
# read data
with open(file_data, 'r') as f:
    keys = [i.strip() for i in f.readline().split("|")]
    data = [ {k:v.strip() for k, v in zip(keys, line.split("|"))} for line in f]
# clean from doped athletes
doped_athletes = ["Tyson Gay", "Yohan Blake", "Tim Montgomery", "Ben Johnson"]
for i, item in enumerate(data):
    if item['Athlete'] in doped_athletes:
        data.remove(item)
# convert time in float
for i in data:
    i['Time'] = float(i['Time'])

for d in data:
    print("{}\t{}\t{}".format(d['Time'], d['Athlete'], d['Nationality']))
```

### Athlete with more records ?

```python
freq_athlete = dict()
for record in data:
    freq_athlete.update({record['Athlete']: freq_athlete.get(record['Athlete'], 0) + 1})

print(f"Il numero massimo di record per un atleta e': {max(freq_athlete.values())} ed è di {[k for k,v in freq_athlete.items() if v == max(freq_athlete.values())][0]}")
```

### Nations with more records ?

```python
freq_athlete = dict()
for record in data:
    freq_athlete.update({record['Nationality']: freq_athlete.get(record['Nationality'], 0) + 1})

print(f"Nazione che ha più record e': {[k for k,v in freq_athlete.items() if v == max(freq_athlete.values())][0]} con {max(freq_athlete.values())} record")
```
