### read from file

```python
f = open("file_to_read", "r") # open file

print(f.read()) # read all file at once

f.close() # close file
```

<div class="alert alert-info">

* `open(name[, mode[, buffering]])` Open file and return a stream.

</div>

**mode** determines the access mode in which the file has opened (read, write, append, etc.).
It is an optional parameter and the default is read (r).

| mode |  |
| :--  | :-- |
| 'r'  |     open for reading (default) |
| 'w'  |     open for writing, truncating the file first |
| 'x'  |     create a new file and open it for writing |
| 'a'  |     open for writing, appending to the end of the file if it exists |
| 'b'  |     binary mode |
| 't'  |     text mode (default) |
| '+'  |     open a disk file for updating (reading and writing) |
| 'U'  |     universal newline mode (deprecated) |

```python
# all mode options:
# 'r'       open for reading (default)
# 'w'       open for writing, truncating the file first
# 'x'       create a new file and open it for writing
# 'a'       open for writing, appending to the end of the file if it exists
# 'b'       binary mode
# 't'       text mode (default)
# '+'       open a disk file for updating (reading and writing)
# 'U'       universal newline mode (deprecated)
```

it is good practice to use `with` statement to manage file.
`with` is used for **context management** (entry, exit and execution)

```python
with open("file_to_read", "r") as f: # with open context
    print(f.read()) # read all file
                    # return a string in text mode
                    # return a bytes object in binary mode
```

`readline()` reads **a line** from file. A newline character `\n` is left at the end of line (except for the last line of the file). 

```python
with open("file_to_read", "r") as f:
    print(f.readline())
    print(f.readline().strip())
    print(f.readline())
```

`readlines()` reads **all lines** of the file and returns a list with all the lines. 

```python
with open("file_to_read", "r") as f:
    print(f.readlines())
```

file object is also **iterable**. <br>
Thus for reading file by line you can loop over file object. <br>
It is a good practice because is fast, memory efficient and simple to read.

```python
with open("file_to_read", "r") as f:
    for line in f:
        print(line)
```

```python
with open("file_to_read", "r") as f:
    for line in f:
        print(line.strip())
```

### write to file

`write()` function writes a **string** to a file returning the number of characters written

```python
with open("file_to_write", "w") as f:
    written = f.write("Don't Panic!\n")

print("characters written =", written)
```

other types need to be converted

```python
with open("file_to_write", "a") as f:
    f.write(str(42))
```

`writelines()` writes all context of a list to file

```python
lines = ["\"So long,\n", "and thanks for all the fish!\"\n",
         "The last ever dolphin message.\n"]

with open("file_to_write", "a") as f:
    f.writelines(lines)
```
