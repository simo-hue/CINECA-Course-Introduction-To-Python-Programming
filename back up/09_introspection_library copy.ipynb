# Introspection

Introspection is an act of **self** examination. \
In computer programming, introspection is the ability to determine the **type** of an object at **runtime**. \
Python provides large support for introspection.

In Python everything is an **object**. Every object may have attributes and methods. \
By using introspection, we can **dynamically inspect Python objects**.

````python
l = [] # take a list
````

`type` function returns the type of an object

````python
type(l)
````

`id` function returns unique ids of an object

````python
id(l)
````

<div class="alert alert-info">

* `isinstance(object, type)` Return whether an object is an instance of a class

</div>

````python
isinstance(l, list)
````

````python
isinstance(l, dict)
````

`dir` function returns a list with attributes and methods of an object.

````python
print( dir(l) )
````

`dir()` without any argument returns all names in the current scope

<div class="alert alert-info">

* `dir([object])` If called without an argument, return the names in the current scope. 
  Else, return an alphabetized list of names comprising (some of) the attributes
  of the given object, and of attributes reachable from it.
    
</div>

`hasattr` function returns `True` if an attribute belongs to an object and `False` otherwise

<div class="alert alert-info">

* `hasattr(object, name)` Return whether the object has an attribute with the given name.

</div>

````python
hasattr(l, 'append')
````

````python
hasattr(l, 'items')
````

````python
hasattr(dict, 'items')
````

`callable` function checks if an object is callable. It returns `True` if the object is callable otherwise `False`

````python
callable(l.append)
````

````python
callable('a string')
````

<div class="alert alert-info">

* `callable(object)` Return whether the object is callable (i.e., some kind of function).

</div>

`getattr` function returns the contest of an attribute if there is some

````python
print( getattr(l, 'append') )
````

````python
print( l.append )
````

<div class="alert alert-info">

* `getattr(object, name[, default])` Get a named attribute from an object; getattr(x, 'y') is equivalent to x.y.
  When a default argument is given, it is returned when the attribute doesn't exist; without it, an exception is raised in that case.

</div>

all methods in `list`

````python
for attr in dir(l):          # loop over all attributes in list
    m = getattr(l, attr)     # get attribute
    if (callable(m)):
        if not attr.startswith('__') and not attr.endswith('__'): # exclude methods starts and ends wit '__'
            print(attr)
````

**docstring** can be inspected with `__doc__` attribute

````python
def f():
    """docstring from function f"""
    pass

print(f.__doc__)
````

````python
print(range.__doc__)
````

documentation of all callable attributes in `list`

````python
for attr in dir(l):              # loop over all attributes in list
    m = getattr(l, attr)         # get attribute
    if (callable(m)):
        if not attr.startswith('__') and not attr.endswith('__'): # exclude methods starts and ends wit '__'
            print("\n>>>", attr)
            print(m.__doc__)     # print attribute docstring
````

`help` function provides informations stored in a module using the doc strings

````python
help(list)
````

## glob - pathname pattern expansion

`glob` module provides a function for making file lists from directory wildcard searches
<br>
it finds all the pathnames matching a specified pattern according to the rules used by the Unix shell
<br>
results are returned in arbitrary order

````python
import glob

ls = glob.glob('*.ipynb')

print(ls)
````

````python
for i in ls[:3]:
    print(i)
    print(os.path.abspath(i))
    print(os.path.dirname(os.path.abspath(i)))
    base, ext = os.path.splitext(i) # Split the pathname path into
                                    # base_name + extension
    print(base, "+", ext)
    print()
````

````python
%%writefile copy_file.py
# %load copy_file.py
#!/usr/bin/env python3
import os
import glob
import shutil
import argparse
import logging

logger = logging.getLogger()
logger.setLevel(logging.DEBUG)

parser = argparse.ArgumentParser(description='copy all file with given extension from input directory to output directory.')

parser.add_argument('-i', '--input-dir',
                    dest='input_dir',
                    default=os.getcwd(),
                    help='Input directory. default %(default)s')

parser.add_argument('-o', '--output-dir',
                    dest='output_dir',
                    default='output_dir',
                    help='Output file. default %(default)s')

parser.add_argument('-e', '--extension',
                    dest='extension',
                    required=True,
                    help='File extension')

args = parser.parse_args()

logging.debug("args %s", args)

if not os.path.exists(args.output_dir):
    logging.info("creating directory %s", args.output_dir)
    os.makedirs(args.output_dir)

ls = glob.glob(os.path.join(args.input_dir, '*.' + args.extension))

logging.info("copy files          : %s", ", ".join([os.path.split(f)[1] for f in ls]))
logging.info("copy from directory : %s", args.input_dir)
logging.info("copy to directory   : %s", args.output_dir)

for f in ls:
    logging.info("coping file %s to directory %s", f, args.output_dir)
    shutil.copy(f, args.output_dir)
````

## CSV File Reading and Writing

````python
import csv
with open('eggs.csv', 'w', newline='') as csvfile:
    spamwriter = csv.writer(csvfile, delimiter=',',
                            quotechar='"', quoting=csv.QUOTE_MINIMAL)
    spamwriter.writerow(['Spam'] * 5 + ['Baked Beans'])
    spamwriter.writerow(['Spam', 'Lovely, Spam', 'Wonderful, Spam'])
````

````python
with open('eggs.csv', 'r', newline='') as csvfile:
    spamreader = csv.reader(csvfile, delimiter=',', quotechar='"')
    for row in spamreader:
        print(row)
````
