# Object Oriented Pillars

Object Oriented Programming (OOP) is a programming approach which describes/expresses a problem or a task as the *interaction* between basic objects which represent the entities in the field contest your program deals with.

The programmer can create new data types (called objects, or classes) which hold together some properties (data members) and functionalities (methods) which can act on themselves properties or interact with other objects' properties.

The programmer can create many variables of these new types (*create instances of an object*). Each instance will maintain its own state and scope.

The pillars of OOP are:
* **encapsulation:** binds together the data and functions that manipulate the data, keeping it safe from outside interference and misuse. internal state data should be *private* (not visible) from other objects outside of itself, and should be changed only through methods of the object to maintains coherence of its representation.
* **abstraction:** objects behaviour (what an object can do, interact with, expose its status) should not deal with the actual internal algorithmic implementation. A *client* which use that object does not need to know the object internal, but just call its methods to use or operate on it. 
* **inheritance:** you can build objects by *assembling* other objects together (such as you can build a bicycle using wheels, pedals, chain, etc) or define a new object starting from a previously defined one and *extend* it with other new properties/methods. The new created object is said to be *derived* from the *base* object
* **polymorphism:** is the ability to specialize the behaviour of a derived object methods for its own purposes and peculiarity of the object class type

See also: https://youtu.be/pTB0EiLXUC8 for a very brief intro on OOP concepts

## Methods in a class

**Function** can be declared in the class body.<br>
This means that the function name is bound within the class namespace.

````python
class Complex:

    def conjugate(c):
        conj = Complex()
        conj.real = c.real
        conj.imag = - c.imag
        return conj
````

````python
x = Complex()

x.real = 42
x.imag = 104

x.real, x.imag
````

````python
y = Complex.conjugate(x)

y.real, y.imag
````

When a reference to a method of class is called from an instance, it is assumed that the Python interpreter knows on which instance the call come from.<br>
Thus the interpreter inserts the **instance itself** as first argument of the function.

````python
y = x.conjugate() # the first argument is implicit to itself

y.real, y.imag
````

As universal convention in Python the first argument of a method should be called **self**.

````python
class Complex:

    def conjugate(self):
        conj = Complex()
        conj.real = self.real
        conj.imag = - self.imag
        return conj
````

## Special methods for arithmetic operators

There are also special methods for **arithmetic operators** like: addition, subtraction, multiplication, division and the absolute value.

```
x + y      x.__add__(self, y)
x - y      x.__sub__(self, y)
x * y      x.__mul__(self, y)
x / y      x.__div__(self, y)
x ** p     x.__pow__(self, y)
abs(x)     x.__abs__(self)
```

````python
class Complex:

    def __init__(self, real, imag=0):
        self.real = real
        self.imag = imag

    def __repr__(self):
        return "{} + {}j".format(self.real, self.imag)

    def __add__(self, other):
        return Complex(self.real + other.real, self.imag + other.imag)
````

````python
x = Complex(42)
y = Complex(12, 5)

x, y
````

````python
x + y
````

## In-place special methods for arithmetic operators

```
x += y     x.__iadd__(self, y)
x -= y     x.__isub__(self, y)
x *= y     x.__imul__(self, y)
x /= y     x.__idiv__(self, y)
```

````python
from math import atan2

class Complex:

    def __init__(self, real, imag=0.0):
        self.real = real
        self.imag = imag
        
    def to_trig(self):
        r = (self.real**2 + self.imag**2)**0.5
        argom = atan2(self.imag, self.real)
        return r, argom

    def __repr__(self):
        return "{} + {}j".format(self.real, self.imag)

    def __add__(self, other):
        return Complex(self.real + other.real, self.imag + other.imag)
    
    def __sub__(self, other):
        return Complex(self.real - other.real, self.imag - other.imag)
    
    def __mul__(self, other):
        a = self.real
        b = self.imag
        c = other.real
        d = other.imag
        
        nuova_parte_reale = (a * c) - (b * d)
        nuova_parte_immaginaria = (a * d) + (b * c)
        
        return Complex(nuova_parte_reale, nuova_parte_immaginaria)

    def __div__(self, other):
        a = self.real
        b = self.imag
        c = other.real
        d = other.imag
        
        nuova_parte_reale = (a * c/(c**2))
        nuova_parte_immaginaria = (b)*(d/(d**2))
        
        return Complex(nuova_parte_reale, nuova_parte_immaginaria)

    def __str__(self) -> str:
        if self.imag >= 0:
            return f"{self.real} + {self.imag}j"
        else:
            return f"{self.real} - {-self.imag}j"
````

````python
x = Complex(42, 3)
y = Complex(2, 1)

print(x.__div__(y))

print(x)

y = Complex(12, 5)

x, y
````

````python
x += y

print(x)

x *= y

print(x)

r1, theta1 = x.to_trig()
print(f"z1 in forma trig: r={r1:.2f}, theta={theta1:.2f} rad")
````

## Inheritance

In object oriented programming inheritance is the capability of one class to derive or inherit the properties from another class.
* Provides **code re-usability**
* Reflects real **relationship**
* Is **transitive**. If a class derived from a base class, all subclasses derived from this class inherits from base class

The class that inherits is called derived class and the one from which it inherits is called base class.

The Python inheritance syntax

````python
class Base:
    pass

class Derived(Base):
    pass
````

## Polygon class example

````python
class Polygon:

    def __init__(self, *args):
        self.sides = args

    def perimeter(self):
        return sum(self.sides)
````

````python
p = Polygon(3, 4, 8, 10)

print("polygon sides number =", len(p.sides))
print("polygon perimeter =", p.perimeter())
````

## Inherit from Polygon class

````python
import math

class Triangle(Polygon):

    def __init__(self, a, b, c):
        Polygon.__init__(self, a, b, c)

    def area(self):
        p = self.perimeter() / 2
        a, b, c = self.sides
        return math.sqrt(p * (p - a) * (p - b) * (p - c))
````

````python
x = Triangle(4, 3, 5)

print("triangle perimeter =", x.perimeter())
print("triangle area =", x.area())
````

* Derived class can expand features of base class.
  Triangle class has area method not present in Polygon class.
* If a requested attribute, like perimeter method in Triangle class, is not found in the derived class, it is looked in the base class, the Polygon class.<br>
  This rule is applied recursively if the base class is derived from other classes.
* Base class methods can called by derived class directly using syntax BaseClass.Method.<br>
  e.g.: in `__init__` method of derived class Triangle is called `__init__` method of base class Polygon

## super() method

In the case we need to call base class methods from derived class:
1. Base class methods can be called directly from derived class
1. A better option is to use built-in function **super()**

## super() method

super() gives access to a method in a base class from a derived class.<br>
In detail super() returns a temporary object of base class from which a base class method can be called.<br>
This is useful for calling inherited methods that have been overridden

There are two typical uses for super.
* super can be used to call a base class method without naming base class explicitly.<br>
  This allow to write more maintainable code and to change base class with very minimal code changes
* Using super allow to support multiple inheritance.<br>
  How to search for inherited methods is defined by the method resolution order (MRO).
  Every class has an `.__mro__` attribute that allows us to inspect the MRO.

## Inherit from Polygon class

````python
class Quadrilateral(Polygon):

    def __init__(self, a, b, c, d):
        super().__init__(a, b, c, d)
        
        
class Square(Quadrilateral):

    def __init__(self, side):
        super().__init__(side, side, side, side)

    def area(self):
        a = self.sides[0]
        return a * a

    def perimeter(self):
        a = self.sides[0]
        return a * 4
````

* Derived class may **override** methods of its base class.
  As in the case of perimeter method of Square class
* Python supports **Multilevel Inheritance**.
  When a class is derived from a class which is also derived from another class.
  Square class derives from Quadrilateral which derives from Polygon class.

## Class introspection

Python has two built-in functions for inheritance introspection:
* **isinstance()** is used to check an instance type. It returns True if the given class is in the inheritance chain of the object class.
* **issubclass()** is used to check class inheritance. It returns True if the first class contains the second class in its inheritance chain.

````python
polygon = Polygon(3, 4, 8, 10)
triangle = Triangle(4, 3, 5)
square = Square(2)
````

````python
isinstance(triangle, Quadrilateral)
````

````python
issubclass(Triangle, Polygon)
````

````python
issubclass(Square, Polygon)
````

## classmethod

A class method:
* is a method which is **bound to a class** and not to an object of the class
* can access and modify the class state
* takes the class as first argument
* is generally used to create factory methods
* is defined with decorator `@classmethod` before function definition

## classmethod

A class method is used to create a polygon from a text file.

````python
class Polygon:

    def __init__(self, *args):
        self.sides = args

    def perimeter(self):
        return sum(self.sides)

    @classmethod
    def create_from_file(cls, filename):
        print(cls)
        with open(filename, "r") as f:
            sides = [int(i) for i in f.read().split()]
        return cls(*sides)
````

## static method

A static method:
* is a method which is **bound to a class** and not to an object of the class
* cannot access or modify the class state
* has not an implicit first argument
* is present in a class because it is related to the class
* knows nothing about class state
* is generally used to create utility functions
* is defined with decorator `@staticmethod` before function definition

## static method

A static method is used to return the sum of internal angle of a polygon given by the formula $180 * (N -2)$ where $N$ is the number of sides.

````python
class Polygon:

    def __init__(self, *args):
        self.sides = args

    def perimeter(self):
        return sum(self.sides)

    @staticmethod
    def internal_angles(n):
        return 180.0 * (n - 2)
````

````python
Polygon.internal_angles(3)
````

````python
Polygon.internal_angles(4)
````

## Python privacy

Private variables **do not exist** in Python.<br>
All attribute in a Python class is public.

All name prefixed with an underscore `_` is conventionally considered private.<br>
It should considered an implementation detail and subject to possible modification without announcement.


In order to avoid name clashes between names defined by a class and by its subclasses there is a mechanism of name mangling for class-private members.<br>
Any name started with at least two leading underscores is textually replaced with prefix `_classname` where classname is the class name with leading underscore stripped.<br>
Name mangling is introduced to allow derived classes to override methods without breaking intraclass method calls.

````python
class A:
    __secret = 42
````

````python
print(A.__secret) ## raises an error
````

````python
print(dir(A))
````

````python
print(A._A__secret)
````
