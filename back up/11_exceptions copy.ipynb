## Catching errors
* Codes should be written to avoid error conditions as much as possible
* Or gracefully handling them informing the user about
   * the type and reasong of fail
   * how to fix the problem
* Even simple functions have many ways to fail

```python
#def inverse_number(a):
#    try:
#        return 1/a
#    except ZeroDivisionError:
#        print("Error: division by zero is not allowed.")
        
def inverse_number(a):
    return 1/a

inverse_number(3.)
inverse_number(0)
```

* Python is very polite about reporting errors but in realistic codes, an error should be not just reported but treated
   * in the jargon: it is necessary to `catch` it
* We already discussed a way to somehow do it but...

```python
def inverse_number(a):
    ret_value = 1/a if a != 0 else None
    return ret_value

print(inverse_number(0))
```

```python
def inverse_number(a):
    ret_value = 1/a if a != 0 else None
    return ret_value
print(inverse_number("a"))
```

* How to deal with such cases?
   * we could add another if
   * or use Python exception handling

```python
def inverse_number(a):
    try:
        return 1/a
    except:
        return None

print(inverse_number("a"))
```

* But this is a bit rude, isn't it? 
   * why did it fail?
* The except clause may specify a variable after the exception name
   * the variable is bound to an exception instance with the arguments stored in instance.args

```python
def inverse_number(a):
    try:
        return 1/a
    except Exception as inst:
        print(inst.args)
        return None

print(inverse_number("a"))
```

* However, we have greatly improved the situation:
   * we detect the exception and can decide if the code must continue or not, and how
* We can do better:
   * exceptions have names and may be catched in different blocks
   * `else` block is executed if no exception has been detected
   * `finally` is executed in any case (even if other blocks return)

```python
def inverse_number(a):
    try:
        result = 1/a
    except ZeroDivisionError:
        print("division by zero!")
        return 1
    except Exception as inst:
        print(inst.args)
        return None
    else:
        print("no exception, returning result")
        return result
    finally:
        print("doing finally it in any case")

print(inverse_number(3), end="\n---\n")
print(inverse_number(0), end="\n---\n")
print(inverse_number("a"), end="\n---\n")
```

## Raising exceptions

* We discussed how to catch exceptions but
   * how to create them? `raise`

```python
def inverse_number(a):
    try:
        result = 1/a
        if result > 10000:
            raise Exception('too large result','> 10000')
    except ZeroDivisionError:
        print("division by zero!")
        raise Exception('wrong number')
    except Exception as inst:
        print(inst.args)
        raise # Exception(inst.args)
    else:
        print("no exception, returning result")
        return result
    finally:
        print("doing finally it in any case")
try:
    inverse_number(0.000000)
except Exception as inst:
    print(inst.args)
```

* Exception in function chains
   * when there is a stack of function calls, exceptions can be handled at each level
   * otherwise the exceptions (intuitively) follow the stack

```python
def inverse_number(a):
    return 1/a

def outer_function(a):
    return inverse_number(a)

outer_function(0)
```

```python
def inverse_number(a):
    try:
        result = 1/a
        if result > 10000:
            raise Exception('too large result','> 10000')
    except ZeroDivisionError:
        print("division by zero!")
        raise Exception('wrong number')
    except Exception as inst:
        print(inst.args)
        raise # Exception(inst.args)
    else:
        print("no exception, returning result")
        return result
    finally:
        print("doing finally it in any case")

def outer_function(a):
    # here the exception is not handled, 
    res = inverse_number(a)
    return res

try:
    r = outer_function(0)
    print(r)
except:
    print("error handled at main unit level")

print("----")
# not catching exception on the outer level
#r = outer_function(0)
```

## Summary 
* it is also possible to create user-defined exceptions (e.g.,  to ZeroDivisionError)
   * not covered in this course
   * first look at the built-in exceptions available
   https://docs.python.org/3/library/exceptions.html
* raising and catching exceptions follow the stack of function call and allows to handle the error at the code layer where it is more appropriate
* beware: try/except is very powerful, do not abuse it as a replacement of doing the right operations
   * but having a program failing in unpredicted way is the big evil

## Hands-on 8.1
* The following code read some numbers -- one for each line -- from a file named `mydata.dat` and compute the average number
   * try to improve it to handle possible expected errors

```python
avg = 0.
real_lines = 0

with open("mydata.dat", "r") as f:
    ls = f.readlines()
    
    #print(f"File Preview: {ls}")
    
    for l in ls:
        try:
            avg += float(l.strip())
            real_lines += 1
        except ValueError as ve:
            print(f"Warning: could not convert {type(l.strip())} to float: '{l.strip()}'. Skipping this line.")
            continue
    
    print("Starting processing the file.")
    try:
        avg /= real_lines
    except ZeroDivisionError:
        print("Error: cannot compute average of empty file.")
    finally:
        print("Finished processing the file.")
print("average value: ",avg)  
```
