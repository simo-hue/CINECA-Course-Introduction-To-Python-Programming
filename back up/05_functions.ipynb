# Functions

### Let's solve second degree equations
$$ ax^2 + bx + c = 0 $$
if $a = 0$
$$ x_1=x_2=-\frac{c}{b} $$
if $a \neq 0$

$$ \Delta = b^2 - 4ac $$

$$ x_{1,2} = \frac{-b \pm \sqrt{\Delta}}{2a}$$

````python
def second_degree_eq_solver(a,b,c):
    print("solving ", a, "x^2 + ", b, "x + ", c, "= 0")

    if a == 0:
        x1 = -c/b
        print("The solution is:",x1)
    else:
        x1 = (-b-(b**2-4*a*c)**0.5)/(2*a)
        x2 = (-b+(b**2-4*a*c)**0.5)/(2*a)
        print("The solutions are: ",x1,x2)

second_degree_eq_solver(1., 3., 2.)
````

* `def` defines a function
* to invoke the function, after the function name the argument list has to be passed
* the function executes its own statements and then exits to the calling unit
* passed variable (actual arguments) are matched to local variables (dummy arguments)

### Returning from a function

````python
def second_degree_eq_solver(a,b,c):
    print("solving ", a, "x^2 + ", b, "x + ", c, "= 0")

    if a == 0:
        x1 = -c/b
        number_of_solutions = 1
    else:
        x1 = (-b-(b**2-4*a*c)**0.5)/(2*a)
        x2 = (-b+(b**2-4*a*c)**0.5)/(2*a)
        number_of_solutions = 2
    
    return number_of_solutions
    
n_solutions = second_degree_eq_solver(1., 3., 2.)
print("number of solutions: ", n_solutions)
````

* A function may `return` a value
   * If no value is returned, implicitly `None` is returned

````python
def second_degree_eq_solver(a,b,c):
    print("solving ", a, "x^2 + ", b, "x + ", c, "= 0")

    x1 = x2 = None
    if a == 0:
        x1 = -c/b
    else:
        x1 = (-b-(b**2-4*a*c)**0.5)/(2*a)
        x2 = (-b+(b**2-4*a*c)**0.5)/(2*a)
    return x1, x2   # <- multiple values returned
    
sol_1, sol_2 = second_degree_eq_solver(0., 3., 2.)
print("solutions are: ", sol_1, sol_2)
````

* A function may also return multiple values just separating them by commas

### Calling function from another function

````python
def delta(a,b,c):
    return b**2-4*a*c

def second_degree_eq_solver(a,b,c):
    print("solving ", a, "x^2 + ", b, "x + ", c, "= 0")

    x1 = x2 = None
    if a == 0:
        x1 = -c/b
    else:
        x1 = (-b-delta(a,b,c)**0.5)/(2*a)
        x2 = (-b+delta(a,b,c)**0.5)/(2*a)
    return x1, x2
    
print("test delta: ", delta(1, 3, 2))
sol_1, sol_2 = second_degree_eq_solver(1., 3., 2.)
print("solutions are: ", sol_1, sol_2)
````

* A function may call another function
   * a good practice is usually to write many functions to split the tasks of the code, improving readability, reducing code replication

### Trying to get solutions from arguments (sort of _pass by reference_)

````python
def second_degree_eq_solver(a,b,c,x1,x2):
    print("solving ", a, "x^2 + ", b, "x + ", c, "= 0")

    def delta_inside(a,b,c):
        return b**2-4*a*c
    
    if a == 0:
        x1 = -c/b
    else:
        x1 = (-b-delta(a,b,c)**0.5)/(2*a)
        x2 = (-b+delta(a,b,c)**0.5)/(2*a)
    
sol_1 = sol_2 = None
second_degree_eq_solver(0., 3., 2., sol_1, sol_2)
print("solutions are: ", sol_1, sol_2) # why is this not working ?
````

* We could think to pass the solutions of equations as additional arguments instead of returning them
* It does not work. Why?
   * Because reassigning scalars inside the function creates new variables (i.e. label to a different object) so that, when exiting the function, the passed variables still have the old values
   * Not possible to fix it for scalars, they are `immutable`!

### Use functions to compute trapezoidal integration of $f$ between $[a,b]$


````python
def trapezoidal_integration(a, b, tol, f, maxsteps=2**23):
    # initialization
    intervals = 1 ; acc = (f(a) + f(b))*0.5 ; dx = (b - a) / intervals
    estimate = acc * dx ; prev_estimate = estimate + 2*tol 
    convergence = False ; steps = 1
    # iterative estimate
    while abs(estimate - prev_estimate) > tol:
        print("Current estimate using ", intervals, " intervals with dx =", dx, " is ", estimate)
        prev_estimate = estimate
        intervals *= 2 ; dx *= 0.5
        for i in range(1, steps + 1, 2): # only contributions from new points
            x = a + i*dx ; acc += f(x)
        estimate = acc*dx
        steps = steps + 1
        if steps > maxsteps: 
            break
    else:
        convergence=True
    return estimate, convergence

def integrand(x):
    return x**2
````

* It is possible to (easily) **pass functions as arguments**
   * our integral function becomes very general
* Selecting functions at run-time is very easy (very powerful feature)

````python
integral, success = trapezoidal_integration(a=-1.0, b=2.0, tol=1.e-4, f=integrand)
print("integral, success: ", integral, success)
````

* Variables outside any function are globally scoped and can be accessed inside the nested functions
   * global variables can easily make your code less readable/maintainable!

````python
a = 4

def do_something():
    print("inside a: ", a)
    
print("outside before a: ",a)
do_something()
print("outside after a: ", a)
````

### Docstrings styles: Google vs NumPy
There are two styles of docstrings: [Google](https://google.github.io/styleguide/pyguide.html#38-comments-and-docstrings) and [NumPy](https://numpydoc.readthedocs.io/en/latest/format.html#docstring-standard). The main difference between the two styles is that Google uses indentation to separate sections, whereas NumPy uses underlines.

#### Google style:

```python
def func(arg1, arg2):
    """Summary line.

    Extended description of function.

    Args:
        arg1 (int): Description of arg1
        arg2 (str): Description of arg2

    Returns:
        bool: Description of return value

    """
    return True
```
#### NumPy style:

```python
def func(arg1, arg2):
    """Summary line.

    Extended description of function.

    Parameters
    ----------
    arg1 : int
        Description of arg1
    arg2 : str
        Description of arg2

    Returns
    -------
    bool
        Description of return value

    """
    return True
```
Google style tends to be easier to read for short and simple docstrings, whereas NumPy style tends be easier to read for long and in-depth docstrings.

The choice between styles is largely aesthetic, but the two styles should not be mixed. Choose one style for your project and be consistent with it. (source: sphinx documentation)

## Python modules
* For realistic projects you have several functions
   * putting them in different files/folders is usually a good option
   * a proper organization is crucial for code maintainability
* The first step is putting functions in different files creating so called Python modules
   * to use the functions in a module you have to `import` it
   * the module name (to be imported) corresponds to the file name
   * a module can host variables too
* When imported from another Python source file, the file name is treated as a namespace (remember the Zen? "*...Namespaces are one honking great idea -- let's do more of those!*")
   * when using the function, the name of the module must be prepended and separated by `.`
* You can import modules at the beginning of the calling unit, usually a good choice
   * or later, to make more readable where it is used
   * or even inside a function

## `import`ing modules
* It is possible to give an alias to the imported module
   * `import <module> as <module_alias>`
* It is also possible (and often recommended) to limit the `import` to a set of specific functions to be imported
   * `from <module> import <function>`
   * function can be used without prepending the module name
   * and you can import all module stuff using `*`
      * not a good practice, you do not have full control of what you are importing in the global namespace

## `import`ing modules - 2

* A module can be imported multiple times from the same code
   * under the hood, only one import is done 
   * no performance loss by importing it again
* Summarizing:
   * `import foo` 
   * `import foo as f`
   * `from foo import bar`
   * `from foo import * # <- do not do it, very bad`

````python
import mymodule  # basic import
print(mymodule.parabola(3))
````

````python
import mymodule as mm # import with renaming
print(mm.parabola(3))
````

````python
from mymodule import parabola, nice_constant
# select functions to import
print(parabola(3))
````

````python
from mymodule import * # select all functions to import
print(parabola(3))
````

## Testing modules

* A very cool feature is the possibility to separately execute a module 
   * it is possible to specify actions to be executed only when the module is directly invoked (very useful for testing)
   * these actions are skipped when the module is imported
   * this can be achieved using a simple conditional
   
   `if __name__ == '__main__':`

````python
def parabola(x):
    return x**2

if __name__ == '__main__':
    # test parabola function
    p = parabola(3.)
    print("check that p returns 9 in this case: ",p)
````
