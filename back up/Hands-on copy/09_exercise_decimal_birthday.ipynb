# decimal birthday

1. compute number of days from your birthday
1. compute your next decimal birthday (how many days to the next decimal round number of days)
   <br>
   eg: if your number of days is 14570 compute when you have the decimal birthday of 15000 days

```python
from datetime import datetime, timedelta

birthdate = datetime(2003, 7, 28)

today = datetime.now()

age = today - birthdate
print("----------------------------- COMPLEANNO CLASSICO ---------------------------------------------------------------------------")
print(f"Age in days: {age.days}")
print(f"Age in years: {age.days // 365}")
print("-----------------------------------------------------------------------------------------------------------------------------")


print("\n\n----------------------------- COMPLEANNO DECIMALE ---------------------------------------------------------------------------")
print(f"Il tuo prossimo compleanno decimale sarà a {len(str(age.days)) + 1} cifre")
#next_decimal_birthday = 14570
next_decimal_birthday = age.days
found = False

while not found:
    if next_decimal_birthday % (10**(len(str(age.days)) - 1)) != 0:
        next_decimal_birthday += 1
    else:
        found = True
    
print(f"Il tuo prossimo compleanno decimale sarà a {next_decimal_birthday} giorni, ovvero il {birthdate + timedelta(days=next_decimal_birthday)} tra {next_decimal_birthday - age.days} giorni")
print("-----------------------------------------------------------------------------------------------------------------------------")
```
