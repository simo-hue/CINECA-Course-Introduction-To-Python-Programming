# Triangle wave example
![triangle wave](Images/triangle_wave.png)

$$
    f(x) = 2 \left\lvert ~
           2 \Bigg( \frac{x}{p} - \left\lfloor \frac{x}{p} + \frac{1}{2}  \right\rfloor \Bigg) \right\lvert
           - p 
$$

```python
from math import floor

def triangle_wave(x, period=1.0):
    """Returns triangle wave function value given x value and period"""
    return 2*abs( 2*( x/period - floor(x/period + 0.5) ) ) - period


start = 0.0
end = 2.0
num_point = 100

dx = (end - start)/num_point

# define x values
x = [start + i * dx for i in range(num_point)]

period = 1

# define y values
y = [triangle_wave(i, period) for i in x]

# plot function
import matplotlib.pyplot as plt

plt.plot(x, y)
plt.xlabel("time")
plt.ylabel("amplitude")
plt.show()
```
