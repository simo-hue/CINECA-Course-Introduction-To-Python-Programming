# Iterables, iterators and generators

## Containers and iterables
* Containers: something that you can use the `in` operator on (you can say if the container contains an object or not)
* Iterable: anything you can loop over with a for loop in Python
   * generally a container is iterable (too complex for a basic course)
   * sometimes a container can be indexed but iterations are not performed by indexing!


```python
a_iterable = [1,2,3]

print(2 in a_iterable)

print(type(a_iterable))

for a_element in a_iterable:
    print(a_element)
    
i = 0
for i in range(3):
    print(a_iterable[i])
```

* Is it possible to loop without containers? Why?
   * never having the complete set of data in memory
   * reducing the number of operations keeping a clean programming style

```python
for p in primes:
    print(p)
```

## Iterators
* An iterator
   * is itself an iterable
   * but is **stateful**, it saves the reached iteration
   * the `next` function allows to get the next element of an iterator
   * can be infinite, otherwise finally `next` returns a stop condition
* When is it used?
   * looping an iterable implicitely creates an iterator from it and execute `next` until iterator consumption
   * the `iter` function allows to explicitely get the iterator used by the iterable
   * this is what looping iterables does:   

```python
a_iterable = [1,2,3]
a_iterator = iter(a_iterable)
print(type(a_iterator))
while True:
    try:
        a_element = next(a_iterator)
        print(a_element)
    except StopIteration:
        break
```

## Iterables and iterators
* An iterator is an iterable
   * but having a state, it is *consumable* which can be unwanted in many circumstances
   * the elements can be retrieved only once

```python
a_iterable = [1,2,3]
a_iterator = iter(a_iterable)
for a_element in a_iterator:
    print(a_element)
for a_element in a_iterator: # this won't be executed
    print(a_element)
```

## Iterables and iterators
* Iterables or iterators: which one to use?
   * For a container, e.g. a sequence, being consumable can be not a plus but a problem
   * Therefore, Python iterates an iterable by implicitely creating an ad-hoc iterator so that the status cannot have consequences

```python
a_iterable = [1,2,3]
for a_element in a_iterable:
    for a_element_bis in a_iterable:
        print(a_element, a_element_bis)
```

* Can you understand this behavior?

```python
a_iterable = [1,2,3]
a_iterator = iter(a_iterable)
for a_element in a_iterator:
    for a_element_bis in a_iterator:
        print(a_element, a_element_bis)
```

## Iterables and iterators
summing up ...
* an **iterator** is an object which implements two special methods (called *iterator protocol*):
  * the `__next__()` method returns one element at a time
  * the `__iter__()` method returns an initialized copy of itself
* an **iterable** is an object we can get and iterator from it:
  * using `iter()` function on an iterable returns an iterator *on* that object
  * the `iter()` function calls `__iter__()` method under the hood

# Combinations

```python
poly = ['circle', 'triangle', 'square',]
poly_comb = itertools.combinations(poly, 2)
for each in poly_comb:
    print(each, end=" | ")
print("\n")
poly_comb = itertools.combinations_with_replacement(poly, 3)
for each in poly_comb:
    print(each, end=" | ")
print("\n")
poly_comb = list(itertools.combinations_with_replacement(poly, 3))
for each in poly_comb:
    print(each, end=" | ")
```

# Cycle

```python
colors = ['red', 'orange', 'yellow']
i = 0
for color in itertools.cycle(colors):
    print(color, end=" | ")
    i += 1
    if i > 40: break
```

# Count

```python
print("# forward count (default):")
for i in itertools.count():
    print(i, end=" ")
    if i >= 100: # otherwise an infinite loop !
        break

print("\n# Backward count (using negative step):")
for i in itertools.count(100,-1):
    print(i, end=" ")
    if i <= 0:
        break
        
```

## Generators
* A very pythonic manner to create iterators
* There is an easy way to produce special iterators called generators not starting from an iterable nor writing a Python custom class
   1. using generator functions
   2. or generator expressions
* Where is the advantage? Looping without containers means:
   * **never having the complete set of data in memory**
   * reducing the number of operations keeping a clean programming style

## Generator functions and `yield`

* `yield` works a bit like `return` because it returns a value but saves the state of the function
   * the next time the function is called, execution continues from where it left off, with the same variable values it had before yielding
* a function yielding something is called a generator function and returns a generator
* as any iterator, a generator can be used by
   * requesting `next` element
   * iterating on it (since an iterator is an iterable)
* beware: after the generator has been consumed, it no more returns elements

```python
def pow_two_generator_function():
    i = 0
    while i < 10:
        yield 2**i
        i += 1

print("Use it as an iterable:")
for i in pow_two_generator_function():
    print(i, end= " ")
else:
    print(" <= finished \n")
    
print("Use it as an iterator")
pow_two_generator = pow_two_generator_function()
print(next(pow_two_generator), end=" ")
print(next(pow_two_generator), end=" ")
for i in pow_two_generator:
    print(i, end=" ")
else:
    print(" <= finished \n")

print(type(pow_two_generator))
```

## Generator expressions

## Using existing generators
* Some Python intrinsic functions already return generators instead of containers:
   * `range` returns a generator (not a list), no problem also for huge numbers
   * `mydict.items()` used when looping overy dictionaries returns a generator, no problems also for huge dictionaries
   * `myfile.readlines()` does NOT return a generator: very dangerous for large files
* Several external libraries
   * return generators for lazy evaluation, read the documentation to avoid mistakes
   * accept generators in addition to common containers

## Advantages of generators
* allow you to create iterators in a very pythonic manner
* allow you to **reduce memory allocation** useful especially for huge data
* iterators allow **lazy evaluation**, only generating the next element of an iterable object when requested. This is useful for very large data sets
* may improve code readability
* Generator Functions are an easy way to write Iterators
* Generator Expressions are a very easy way to write Iterators (for simple cases only)

## Working with generators
* An example of task implemented via generators. Can you follow the instruction flow?
* How to make them infinite?

```python
def check_prime(number):
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True

def Primes(maxn):
    number = 1
    while number < maxn:
        number += 1
        if check_prime(number):
            yield number

primes = Primes(10**20)
print(primes)
for x in primes:
    print(x, end=" ")
    if x > 100: 
        break
        
print()
primes = (i for i in range(2, 10**20) if check_prime(i))
print(primes)
for x in primes:
    print(x, end=" ")
    if x > 100: 
        break
```

```python
tg = (i**2 for i in range(10**100))

def doubler(g):
    for ig in g:
        if ig%3 == 0:
            yield ig*2
        else:
            yield ig

def halver(g):
    for ig in g:
        if ig%2 == 0:
            yield ig//2
    
tg_double = doubler(tg)
tg_double_half = halver(tg_double)    

for t in tg_double_half:
    print(t, end=" ")   
    if t > 10000:
        break 
```

## The object point of view
* Iterables have defined an `__iter__` method which returns the corresponding iterator
   * the method is automatically called when looping over an iterable
   * or manually called using the `iter` function
* Iterators have defined
   * an `__iter__` method which must return the iterator itself (but can do other things, e.g. restart consumption counter)
   * a `__next__` method which defines how to return the next element and, optionally, when stop raising a `StopIteration` exception
* Generators are particular iterators which can be defined using generator functions or generator expressions

```python
class PowTwoIterable:
    """Class to implement an iterable of powers of two"""
    
    def __init__(self, max_i = 0):
        self.max_i = max_i
        
    def __iter__(self):
        return PowTwoIterator(self.max_i)        
```

```python
class PowTwoIterator:
    """Class to implement an iterator of powers of two"""

    def __init__(self, max_i = 0):
        self.n = 0
        self.max_i = max_i

    def __iter__(self):
        self.n = 0 # WHY ?
        return self

    def __next__(self):
        if self.n <= self.max_i:       
            result = 2 ** self.n
            self.n += 1
            return result       
        else:
            raise StopIteration      
```

```python
pt = PowTwoIterable(5)        
for i in pt:
    print(i, end=" |A| ")
print("\n")

for i in pt:
    print(i, end=" |B| ")
print("\n")
pt = PowTwoIterator(5)        
for i in pt:
    print(i, end=" |C| ")
print("\n")

for i in pt:
    print(i, end=" |D| ")
```
