# Triangle wave example
![triangle wave](Images/triangle_wave.png)

$$
    f(x) = 2 \left\lvert ~
           2 \Bigg( \frac{x}{p} - \left\lfloor \frac{x}{p} + \frac{1}{2}  \right\rfloor \Bigg) \right\lvert
           - p 
$$

```python
from math import floor

def triangle_wave(x, period=1.0):
    """Returns triangle wave function value given x value and period"""
    return 2*abs( 2*( x/period - floor(x/period + 0.5) ) ) - period


start = 0.0
end = 2.0
num_point = 100

dx = (end - start)/num_point

# define x values
x = [start + i * dx for i in range(num_point)]

period = 1

# define y values
y = [triangle_wave(i, period) for i in x]

# plot function
import matplotlib.pyplot as plt

plt.plot(x, y)
plt.xlabel("time")
plt.ylabel("amplitude")
plt.show()
```

Implement a `smooth` function that smooth the triangle wave.
<br>
A moving average of amplitude `k` points should be performed.
<br>
The value of a given point is computed by averaging from `k` point before to `k` point after.

$$
x_{i} = \frac{1}{2 k + 1} \sum_{j=-k}^{+k} x_{i+j}
$$

```python
def smooth(x, k):
   y = []
   const = 1/(2*k + 1)
   
   for i, item in enumerate(x):
      y.append(sum(x[i-k:i+k+1]) * const)
      
   return y
```

## Plot Animato Approssimazione

```python
import matplotlib.pyplot as plt
from time import sleep
from IPython.display import clear_output, display

# Assumiamo che x, y e la funzione smooth (corretta) siano già definiti
k = 50 

while k > 2:
    k -= 2  # Scendiamo di 2 alla volta per rendere l'animazione più veloce
    
    # 1. Calcola la nuova curva
    y_smooth = smooth(y, k) # Assicurati di usare la versione corretta di smooth!

    # 2. Pulisci l'output precedente (wait=True riduce lo sfarfallio)
    clear_output(wait=True)
    
    # 3. Crea il grafico
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, color='lightgray', label='Dati originali') # Sfondo grigio
    plt.plot(x, y_smooth, color='red', linewidth=2, label=f'Smooth (k={k})')
    plt.legend()
    plt.title(f"Animazione Approssimazione: k = {k}")
    
    # 4. Mostra il grafico e metti in pausa
    plt.show() 
    sleep(0.01) # Pausa di 0.1 secondi tra un frame e l'altro
```

## Plot Singolo Approssimazione

```python
k = 50

y_smooth = smooth(y, k)

plt.plot(x, y)
plt.plot(x, y_smooth)

plt.show()
```
